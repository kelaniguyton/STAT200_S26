<font color=red>Task 2.1</font>

* Write a `for` loop that loads each file in the "Data Scientist" folder and stores it into an unstructured list

  Pseudo code hint:
  ```
  # create empty list
  data <- list()

  # loop through files
  for (i in 1:number_of_files) {
    # store as entry in list
    data[[i]] <- read.file...
  }
  ```


````
# current working directory
data_dir <- "/content/salary_data"

# saving vector with all of the csv file names that are in the "Data Scientist" folder
file_list <- list.files(file.path(data_dir, "Data Scientist"))

# empty list to store data from csv files
data <- list()

# loop through the files addings to my unstructured list `data`
for(i in 1:length(file_list)) {
  # stor as entry in the list
  data[[i]] <- read.csv(file.path(data_dir, "Data Scientist", file_list[i]), header = TRUE)
}

data
````
