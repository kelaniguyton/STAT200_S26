####<font color=red>Task 2</font> - Boolean Operators and Order of Execution



<font color=red>Task 2.1</font>

* What is the result of the following Boolean expression?

  `!FALSE | TRUE & FALSE`

* Explain the order in which the Booleans are evaluated in a text cell.

````
!FALSE | TRUE & FALSE
````

The `!FALSE` is first evaluated to `TRUE`, resulting in `TRUE | TRUE & FALSE`.

The AND operator is then evaluated, resulting in `TRUE | FALSE`.

The OR operator is then evaluated, resulting in `TRUE`.

<br>

<font color=red>Task 2.2</font>

* What is the result of the following Boolean expression?

  `(!FALSE | TRUE) & FALSE`

* Explain the order in which the Booleans are evaluated in a text cell.

````
(!FALSE | TRUE) & FALSE
````

The `!FALSE` is first evaluated to `TRUE`, resulting in `(TRUE|TRUE) & FALSE`.

The expression in the parenthesis is then evaluated, resulting in `TRUE & FALSE`.

The AND operator is then evaluated, resulting in `FALSE`.

####<font color=red>Task 3</font> - Subsetting with Boolean Operators



<font color=red>Task 3.1</font>

* What is the average salary of individuals who have a BS degree AND are less than 25 years of age?

* What is the average salary of individuals who have a PhD degree AND are less than 25 years of age?

* In your opinion, are these salaries very different? Why? Why not?

````
# have BS degree & less than 25 years old
mean(salary$Salary[salary$Education.Level == "BS" & salary$Age < 25])

# have a PhD degree & less than 25 years old
mean(salary$Salary[salary$Education.Level == "PhD" & salary$Age < 25])
length(salary$Salary[salary$Education.Level == "PhD" & salary$Age < 25])
````
