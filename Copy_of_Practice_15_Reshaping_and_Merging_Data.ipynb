####<font color=red>Task 1</font>


<font color=red>Task 1.1</font>

* Import the global life expectancy data from this link:
  https://raw.githubusercontent.com/khasenst/datasets_teaching/refs/heads/main/gapminder_life_expectancy.csv

* Verify the dataset was imported correctly using the `head()` function

  Note the life expectancy is reported in years.

```
# import data
life_expectancy <- read.csv("https://raw.githubusercontent.com/khasenst/datasets_teaching/refs/heads/main/gapminder_life_expectancy.csv", header = TRUE)

# check
head(life_expectancy)
```

<br>

<font color=red>Task 1.2</font>

* Convert the data frame on life expectancy to wide format such that

  * Each country is included as a single row
  * Life expectancy for each year is represented by columns

```
life_expectancy_wide <- reshape(data = life_expectancy,
                                timevar = "year",
                                idvar = "country",
                                direction = "wide")

# check reformating
head(life_expectancy_wide)
```

<br>

<font color=red>Task 1.3</font>

* Using the wide-formatted data frame, determine if the life expectancy of any countries ***decreased*** between the years 2010 and 2015

* Write your observations in a text cell

```
# subtract life expectancy in 2010 from life expectancy in 2015
difference <- life_expectancy_wide$life_expectancy.2015 - life_expectancy_wide$life_expectancy.2010

# subset rows such that the difference in life expectancy is negative (life expectancy went down)
life_expectancy_wide[which(difference < 0), ]
```

There are 5 countries with a decrease in life expectancy between the years 2010 and 2015. This includes Libya, Syria, UAE, Venezuela, and Yamen.

<br>

<font color=red>Task 1.4</font>

* Use `R` code to determine which country had the greatest ***decrease*** in life expectancy?

* Why does this country have such a large decrease in life expectancy? You may search for your answer online

```
# finding country with largest decrease
life_expectancy_wide[which.min(difference), ]
```

Syria has the largest drop in life expectancy between the years 2010 and 2015 due to the Syrian Civil War.

<br>

<font color=red>Task 1.5</font>

* Use `R` code to determine which country had the greatest ***increase*** in life expectancy?

* Why does this country have such a large increase in life expectancy? You may search for your answer online

```
# finding country with largest decrease
life_expectancy_wide[which.max(difference), ]
```

Haiti had the largest increase in life expectancy between the years 2010 and 2015. This was due to the severe drop in life expectancy after the major earthquake, which occured in 2010.

<br>

<font color=red>Task 1.6</font>

* Convert the wide-formatted life expectancy data frame back to long format

* Order the data frame by year and country

* Remove the row names

```
# return to long format
le_long <- reshape(data = life_expectancy_wide,
                   varying = names(life_expectancy_wide)[2:ncol(life_expectancy_wide)],
                   idvar = "country",
                   v.names = "life expectancy",
                   timevar = "year",
                   times = 1960:2016,
                   direction = "long")

# check conversion to long format
head(le_long)
```

```
# order first
le_long <- le_long[order(le_long$year, le_long$country), ]

# remove row names
rownames(le_long) <- NULL

# check our changes
head(le_long)
```

####<font color=red>Task 2</font>


<font color=red>Task 2.1</font>

* Import the global fertility data from this link:
  https://raw.githubusercontent.com/khasenst/datasets_teaching/refs/heads/main/gapminder_fertility.csv

* Verify the dataset was imported correctly using the `head()` function

  Note that fertility is reported in average number of children per woman.

```
# import data
fertility <- read.csv("https://raw.githubusercontent.com/khasenst/datasets_teaching/refs/heads/main/gapminder_fertility.csv", header = TRUE)

# check datasets
head(fertility)
```

<br>

<font color=red>Task 2.2</font>

* Merge the life expectancy data (long format) from Task 1 with the fertility data from Task 2.1 using a ***left join***.

  Note that in our "by" arguments, we must specify both country and year as vector. The merge should result in 10,545 rows.

```
# left join
le_fert <- merge(x = life_expectancy,
                 y = fertility,
                 by.x = c("country", "year"),
                 by.y = c("country", "year"),
                 all.x = TRUE,
                 all.y = FALSE)    # left join so all.x = TRUE and all.y = FALSE

nrow(le_fert)
```

<br>

<font color=red>Task 2.3</font>

* Convert the merged data frame from Task 2.2 into wide format such that
  * Each country is included as a single row
  * Life expectancy and fertility for each year is represented by columns

```
le_fert_wide <- reshape(data = le_fert,
                        timevar = "year",
                        idvar = "country",
                        direction = "wide")

# check
head(le_fert_wide)
```

<br>

<font color=red>Task 2.4</font>

* Plot a scatterplot with fertility during 2015 on the x-axis and life expectancy during 2015 on the y-axis

* In a text cell, write your observations. Is there a trend? Speculate as to why/who not.

  Hint: `plot(x, y)`

```
# scatterplot
plot(le_fert_wide$fertility.2015,
     le_fert_wide$life_expectancy.2015,
     xlab = "Fertility (number of children per women)",
     ylab = "Life Expectancy (years)",
     main = "2015")
```

There appears to be a negative correlation between life expectancy and fertility. Countries with higher life expectancy tend to have fewer children, on average per woman. This could be attributed to several factors, including the effect of birth on longevity and relationship between fertility and socioeconomic factors.

<font color=red>Task 3.2</font>

* Merge the geography dataset with the merged data from Task 2.3 using an ***inner join***

```
# inner join
le_fert_geo <- merge(x = geography,
                     y = le_fert,
                     by.x = "country",
                     by.y = "country",
                     all.x = FALSE,
                     all.y = FALSE)

# check
le_fert_geo
```

```
# fertility & continent
boxplot(le_fert_geo$fertility ~ le_fert_geo$continent)

# life expectancy & continent
boxplot(le_fert_geo$life_expectancy ~ le_fert_geo$continent)
```
