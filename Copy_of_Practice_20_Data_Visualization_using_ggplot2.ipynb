<font color=red>Task 2.1</font>

* Plot a histogram of the `life_expectancy` variable with `bins = 100` using `ggplot()`

* Include labels for the title, x-axis, and y-axis

* What are your observations? For example, based on the histogram
  * What is the average life expectancy?
  * What is the range of life expectancy?
  * Is the distribution unimodal? Skewed?
  * etc.

```
# start with our base ggplot()
ggplot(data = gapminder, aes(x = life_expectancy)) +
  geom_histogram(bins = 100) +
  labs(title = "Histogram of Life Expectancy",
      x = "Life Expectancy (Years)",
      y = "Count")
```

The average life expectancy appears to be ~ 72 years old. The range of life expectancy appears to be between the ages of ~ 13 to 85 years old. The distribution is unimodal and left skewed.

```
gapminder$life_expectancy_highlow <- ifelse(gapminder$life_expectancy > 70, "high", "low")
```

The script creates a new variable in the `gapminder` data frame called `life_expectancy_highlow`, which indicates if an observation has a life expectancy above 70 (high) or below 70 (low).

<br>

<font color=red>Task 5.2</font>

* Plot a heatmap with `year` on the y-axis and `life_expectancy_highlow` on the x-axis using `ggplot()`.

* "Add" the layer below to change the color of your plot

  `scale_fill_gradient(low = "white", high = "red")`

* Include labels for the title, x-axis, and y-axis

* What trend do you observe over time? Can you explain the trend you observe?

```
freq_table <- as.data.frame(table(gapminder$year, gapminder$life_expectancy_highlow))
names(freq_table) <- c("year", "life_expectancy_highlow", "Count")

ggplot(freq_table, aes(x = life_expectancy_highlow, y = year, fill = Count)) +
  geom_tile() +
  scale_fill_gradient(low = "white", high = "red") +
  labs(title = "Heatmap of Life Expectancy by Year",
  x = "Life Expectancy",
  y = "Year",
  fill = "Count")
```

The number of observations with life expectancy greater than 70 years is increasing over time.

####<font color=red>Task 6</font>

<font color=red>Task 6.1</font>

* Subset the gapminder dataset to only include the `year` 2000 and name this new data frame `gapminder2000`

* Create a scatterplot with life expectancy on the y-axis and fertility on the x-axis using `ggplot()`

* Include labels for the title, x-axis, and y-axis

* What are your observations?

```
gapminder2000 <- gapminder[gapminder$year == 2000, ]

ggplot(data = gapminder2000, aes(x = fertility, y = life_expectancy)) +
  geom_point() +
  labs(title = "Scatterplot of Life Expectancy vs Fertility",
  x = "Fertility",
  y = "Life Expectancy")
```

Countries with higher life expectancy have lower fertility rates.

<font color=red>Task 6.2</font>

* Include `continent` as a shape visual dimension to your plot from Task 7.1

```
ggplot(data = gapminder2000, aes(x = fertility, y = life_expectancy, shape = continent)) +
  geom_point() +
  labs(title = "Scatterplot of Life Expectancy vs Fertility",
  x = "Fertility",
  y = "Life Expectancy")
```

<br>

<font color=red>Task 6.3</font>

* Choose two other variables and include as a color and size visual dimension in your plot from Task 6.2

* What are your observations?

```
options(repr.plot.width = 7, repr.plot.height =7)
ggplot(data = gapminder2000, aes(x = fertility,
                                y = life_expectancy,
                                color = infant_mortality,
                                size = population,
                                shape = continent)) +
  geom_point() +
  labs(title = "Scatterplot of Life Expectancy vs Fertility",
  x = "Fertility",
  y = "Life Expectancy",
  size = "Population",
  color = "Infant Mortality per 1000",
  shape = "Continent")
```

From the plot, it appears that most countries with high fertility rates, low life expectancies, and high infant mortality are in Africa. There are two squares (Asia) with very large population, which are likely China and India.
